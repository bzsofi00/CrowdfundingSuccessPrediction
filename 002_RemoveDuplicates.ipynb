```python
# import necessary libraries
import pandas as pd
```

```python
# Read the cleaned CSV file into a DataFrame
combined_clean_df = pd.read_csv(r"C:\Code\kickstarter_combined_clean_no_faulty.csv")
```

```python
# Print the shape of the cleaned DataFrame
print("Cleaned DataFrame shape:", combined_clean_df.shape)
```

```python
# number of duplicates in combined_clean_df based on 'id' column (all rows that belong to a duplicated id, including the first occurrence)
duplicates_combined = combined_clean_df[combined_clean_df.duplicated(subset=['id'], keep=False)]
print("Number of duplicate projects in combined_clean_df:", duplicates_combined.shape[0])
```

```python
# number of duplicates using id in combined_clean_df (counts only the extra rows, excluding the first occurrence of each duplicated id)
print("Number of duplicate rows in combined_clean_df using id:", combined_clean_df.duplicated(subset='id').sum())
```

```python
# drop duplicate extra rows in combined_clean_df based on 'id' column, keeping the first occurrence
combined_clean_df = combined_clean_df.drop_duplicates(subset=['id'], keep='first').reset_index(drop=True)
print("Combined clean DF shape after removing duplicates:", combined_clean_df.shape)
```

```python
# show the number of projects
print("Total number of projects:", len(combined_clean_df))
print(combined_clean_df['state'].value_counts())
```

```python
# delete rows where state is 'submitted', 'canceled', 'live', 'suspended', 'started'
irrelevant_states = ['submitted', 'canceled', 'live', 'suspended', 'started']
combined_clean_df = combined_clean_df[~combined_clean_df['state'].isin(irrelevant_states)]

# print number of rows after filtering
print("Number of rows after filtering:", combined_clean_df.shape[0])
print(combined_clean_df['state'].value_counts())
```

```python
# list column names
print("Column names:", combined_clean_df.columns.tolist())
```

```python
# missing values without truncation
pd.set_option('display.max_rows', None)  # Disable truncation of rows
missing_values = combined_clean_df.isnull().sum()
print("Missing values in each column:\n", missing_values)
pd.reset_option('display.max_rows')  # Reset to default after printing
```

```python
# remove irrelevant columns
irrelevant_columns = [
    "photo_key", "photo_full", "photo_ed", "photo_med", "photo_little", "photo_small", "photo_thumb",
    "creator_avatar_small", "creator_avatar_medium", "creator_urls_api_user",
    "category_id", "category_analytics_name", "category_parent_id", "category_position", "category_color", "category_urls_web_discover",
    "creator_chosen_currency", "creator_is_ksr_admin",
    "currency_symbol", "currency_trailing_code", "current_currency", "usd_type", 
    "category_slug", "country_displayable_name",
    "creator_name", "creator_slug", "creator_avatar_thumb", "creator_urls_web_user",
    "fx_rate", "is_launched",
    "location_id", "location_slug", "location_short_name", "location_displayable_name",
    "location_localized_name", "location_is_root", "location_expanded_country",
    "location_urls_web_discover", "location_urls_web_location", "location_urls_api_nearby_projects",
    "photo_1024x576",
    "profile_background_color", "profile_text_color",
    "profile_link_background_color", "profile_link_text_color", "profile_link_text", "profile_link_url",
    "profile_background_image_opacity",
    "usd_exchange_rate",
    "profile_feature_image_attributes_id", "profile_background_image_attributes_id",
    "profile_background_image_attributes_image_urls_default", "profile_background_image_attributes_image_urls_baseball_card",
    "profile",
    "location",
    "creator",
    "creator_id",
    "creator_is_superbacker",
    "creator_is_email_verified",
    "creator_is_registered"
]

combined_clean_df = combined_clean_df.drop(columns=irrelevant_columns, errors='ignore')
print("Remaining columns:", combined_clean_df.columns.tolist())
for col in combined_clean_df.columns:
    print(col)

# number of columns
print("Number of columns after dropping irrelevant ones:", len(combined_clean_df.columns))
print("Number of columns removed:", len(irrelevant_columns))
```

```python
# save the cleaned DataFrame to a new CSV file
combined_clean_df.to_csv(r"C:\Code\kickstarter_combined_clean_no_faulty_no_duplicates.csv", index=False)
```
